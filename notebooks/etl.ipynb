```python
import pandas as pd

# Create sample usuarios.csv
sample = pd.DataFrame({"id": [1,2], "nome": ["Ana","Bruno"], "saldo": [1500, 500]})
sample.to_csv("usuarios.csv", index=False)

sample.head()
```

```python
from src.etl import run_etl

try:
    run_etl("usuarios.csv", "clientes_com_recomendacao.csv")
    print("ETL concluído com sucesso")
except Exception as e:
    print("Erro ao executar ETL:", e)
```
